#1. Username Validator — Strings

Write a program to validate a username.

A username is valid if:

Minimum 5 characters
Contains only letters and digits
Must not contain spaces

```python
username = input("Enter username: ")
username.isdigit()

if len(username) >= 5 and username.isalnum() and " " not in username:
  print("Valid username")
else:
  print("Invalid username")
```

# 5. ATM Withdrawal — Nested if

```python
Balance = 10000

Ask the user for withdrawal amount.

Check:

Amount must be greater than 0
Amount must be a multiple of 100
Amount must not exceed balance

Display:

Withdrawal Successful
Remaining Balance: 7500

Otherwise display the appropriate error message.

Topic: Nested if + arithmetic
```

```python
balance = 10000
amount = int(input("Enter withdrawal amount: "))
if amount > 0:
  if amount % 100 == 0:
    if amount <= balance:
      balance -= amount
      print("Withdrawal successful")
      print("Remaining balance: ", balance)
    else:
      print("Insufficient balance")
  else:
    print("Amount must be  a multiple of 100")
else:
  print("amount must be greater than 0")
```

# 6. Login System — Multiple Conditions

```python

Create a login program.

Correct credentials:

username = "admin"
password = "Python123"

Ask the user for username and password.

Display:

Login Successful

or

Invalid Username or Password

Give the user maximum 3 attempts.

Topic: while loop + conditions
```

```python
correct_username = "admin"
correct_password = "Python123"
attempts = 0
while attempts < 3:
  username = input("Enter your username: ")
  password = input("Enter your password: ")
  if username == correct_username and password == correct_password:
    print("Login successful")
    break
  else:
    print("Invalid username and password")
  attempts += 1
if attempts == 3:
  print("your account is temporarly locked please try after 24hr")
```

#7. Shopping Cart — List + Loop

```python

Create a list of prices:

prices = [500, 1200, 300, 800, 1500]

Calculate:

Total price
Number of products
Highest price
Lowest price
Average price

Example:

Total: 4300
Number of products: 5
Highest: 1500
Lowest: 300
Average: 860

Topic: Lists + loops
```

```python
prices = []
for i in range(5):
  price = int(input("Enter your price: "))
  prices.append(price)
print(prices)
print("total:", sum(prices))
print("Number of product: ", len(prices))
print("Highest: ", max(prices))
print("Lowest: ", min(prices))
print("average: ", sum(prices)/len(prices))
```

# 8. Word Analyzer — String + Loop

```python


Take a word from the user.

Count:

Number of vowels
Number of consonants
Number of digits
Number of spaces

Example:

Enter text: Python123

Vowels: 1
Consonants: 5
Digits: 3
Spaces: 0

Topic: Strings + loops + isalpha(), isdigit()
```

```python
word =  input()
vowels = 0
consonents = 0
Digits = 0
spaces = 0
for ch in word:
  if ch in "aeiouAEIOU":
     vowels += 1
  elif ch >= "a" and ch <= "z":
    consonents += 1
  elif ch >= "1" and ch <= "9":
    Digits += 1
  elif ch == " ":
    spaces += 1
print("Vowels: ", vowels)
print("Consonents: ", consonents)
print("Digits: ", Digits)
print("Spaces: ", spaces)
```

```python
#Method 2
word = input()
vowels = 0
consonents = 0
Digits = 0
spaces = 0
for ch in word:
  if ch in "aeiouAEIOU":
    vowels += 1
  elif ch.isalpha():
    consonents += 1
  elif ch.isdigit():
    Digits += 1
  elif ch == " ":
    spaces += 1
print("Vowels: ", vowels)
print("Consonents: ", consonents)
print("Digits: ", Digits)
print("Spaces: ", spaces)
```

#10. Number Guessing Game — Random + Loop

```python
Generate a random number between 1 and 100.

Ask the user to guess it.

If the guess is:

Too high → Too High
Too low → Too Low
Correct → Congratulations!

Allow maximum 7 attempts.

Example:

Guess: 50
Too Low

Guess: 75
Too High

Guess: 63
Congratulations!

Topic: random module + while loop
```

```python
import random
secret_number = random.randint(1,100)
attempt = 0
while attempt < 7:
  guess = int(input("guess: "))

  if guess > secret_number:
    print("Too high")
  elif guess < secret_number:
    print("Too low")
  else:
    print("congratulation!")
    break
  attempt += 1
if attempt == 7:
  print("Game over")
  print("The number was: ", secret_number)
```
