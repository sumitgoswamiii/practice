```python
Problem 1 : Student Grade Analyzer
Problem Statement
Write a Python program that takes marks of 5 subjects as input.
Your program should:

Store the marks in a list.
Calculate the average marks.
Find the highest and lowest marks.
Display the grade using:
| Average | Grade |
| --------  | ----- |
| 90+        | A |
| 75-89      | B |
| 60-74      | C |
| Below 60   | D |
```

```python
marks = []
for i in range(5):
  m = float(input())
  marks.append(m)

s = sum(marks)
n = len(marks)

avg = s/n

if avg > 90:
  print("Grade A")
elif avg >= 75:
  print("Grade B")
elif avg >= 60:
  print("Grade C")
else:
  print("Grade D")

print("maximum number in a list is =", max(marks),",","minimum number in a list is", min(marks))

print("Average is :", avg)
```

```python
Problem 4 : Character Counter
Problem Statement
Take a string as input from the user.
Display
Number of vowels
Number of consonants
Number of digits
Number of spaces
```

```python
# Ip1: Python3 is Fun
# op1: vowels - 3
# consonants - 8
# digits - 1
# spaces - 2
```

```python
s = input()
s1 = s.lower()
vowels = 0
consonents = 0
digits = 0
spaces = 0
for c in s1:
  if c in "aeiou":
    vowels += 1
  elif c > "a" and c < "z":
    consonents += 1
  elif c > "0" and c < "9":
    digits += 1
  elif c == " ":
    spaces += 1
print("vowels are -", vowels)
print("consonents are -", consonents)
print("digits are -", digits)
print("spaces are -", spaces)
```

```python
Problem 6 : Password Validator
Problem Statement
A password is valid if it contains
minimum 8 characters
at least one uppercase letter
at least one lowercase letter
at least one digit
Display
Strong Password
or
Weak Password
```

```python
password = input("Enter your password: ")
has_uppercase = False
has_lowercase = False
has_digit = False
for ch in password:
  if ch.isupper():
    has_uppercase = True
  elif ch.islower():
    has_lowercase = True
  elif ch.isdigit():
    has_digit = True
if len(password) >= 8 and has_uppercase and has_lowercase and has_digit:
  print("Strong password")
else:
  print("Weak password")
print(has_uppercase)
print(has_lowercase)
print(has_digit)
```
